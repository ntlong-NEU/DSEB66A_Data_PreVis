```python
url1 = 'https://raw.githubusercontent.com/jakevdp/data-USstates/master/state-population.csv'
url2 = 'https://raw.githubusercontent.com/jakevdp/data-USstates/master/state-areas.csv'
url3 = 'https://raw.githubusercontent.com/jakevdp/data-USstates/master/state-abbrevs.csv'

pop = pd.read_csv(url1)
areas = pd.read_csv(url2)
abbrevs = pd.read_csv(url3)
```

```python
df01 = pd.merge(pop,abbrevs,
         left_on="state/region",
         right_on='abbreviation',
         how='inner')
```

```python
df01.head()
```

```python
areas.head()
```

```python
df02 = pd.merge(df01,areas)
df02.head()
```

```python
df03 = df02[['state','ages','year','population','area (sq. mi)']]
df03.head()
```

```python
df03['density'] = df03['population']/df03['area (sq. mi)']
```

```python
df03.head()
```

```python
df04 = df03[(df03['year']==2010)&(df03['ages']=='total')][['state','density']]
```

```python
df04.head()
```

```python
df04['Rank'] = df04['density'].rank(ascending=False)
df04.head()
```

```python
df04.sort_values(by='Rank')
```
