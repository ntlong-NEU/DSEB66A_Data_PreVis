## Lecture 3: Data Aggregation and Group Operations

```python
import numpy as np
import pandas as pd
import seaborn as sns
```

```python
sns.get_dataset_names()
```

```python
planets = sns.load_dataset('planets')
planets.head()
```

```python
df = pd.DataFrame({'key': list("ABCABC"),
                   "data":range(6)},
                   columns=['key','data'])
df 
```

```python
list(df.groupby('key'))
```

```python
for ob in df.groupby('key'):
    print(ob[1])
```

```python
df.groupby('key').sum()
```

```python
planets.duplicated().sum()
```

```python
planets.drop_duplicates(inplace=True)
```

```python
planets[['method','number']].groupby('method').sum().sort_values(by='number',ascending=False)
```

```python
planets[['year','mass','number','orbital_period']].info()
```

```python
# groupby by year:
# sum of number; means of mass, min and max of orbital_period 
planets[['year','mass','number','orbital_period']].groupby('year').aggregate({'mass':'mean',
                                                                              'number':'sum',
                                                                              'orbital_period':['min','max']})
```

```python
planets[['method','number']].groupby('method').sum()
planets.groupby('method')[['number']].sum()
```

```python
planets.groupby('method')[['number']].sum()
```

```python
planets.groupby('method').filter(lambda x: x['number'].sum()>=20)
```
